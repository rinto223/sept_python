```python
# recurssion : afunction calling itself
#write a programe to find factorial of a number using recurssion
n=int(input("enter the num:"))
def fact (n):
  if n==1:
    return 1
  else:
    return n*fact(n-1)
fact(n)

```

```python
#wap to find sum of first n natural number using recurrsion
n=int(input("enter the numbers"))
def sum (n):
  if n==1:
    return 1
  else:
    return n+sum(n-1)
sum(n)
```

```python
#create a class of rectangle and create methods to dipaly its area and perimeter

class Rectangle():
  def get_data(self):
    self.lenght=int(input("enter the lenght:"))
    self.bredth=int(input("enter the breadth"))
  def display_data(self):
    print(self.lenght)
    print(self.bredth)

  def area(self):
    print(self.lenght*self.bredth)

  def perimeter(self):
    print(2*(self.lenght+self.bredth))

r1=Rectangle()
r1.get_data()
r1.display_data
r1.area()
r1.perimeter()


```

```python
#create a class of circles and print area and perimeter using methods
class circle():
  def get_data(self):
    self.r=int(input("enter the radius"))

  def display_data(self):
    print(self.r)
  def area(self):
    print(3.14*self.r*self.r)

  def perimeter (self):
    print(2*3.14*self.r)

c1=circle()

c1.get_data()
c1.area()
c1.perimeter()
```
